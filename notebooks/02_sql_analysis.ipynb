```python
import sqlite3
import pandas as pd

df = pd.read_csv('../data/processed/creditcard_with_features.csv')
conn = sqlite3.connect('../data/fraud.db')
df.to_sql(
    'transactions',
    conn,
    if_exists='replace',
    index=False
)

print("✅ Loaded into SQLite — table: transactions")
```
